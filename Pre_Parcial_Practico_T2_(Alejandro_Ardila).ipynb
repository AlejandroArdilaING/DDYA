# Problema

Queue using two stacks

## Análisis de complejidad

O(n)


## Codigo


```python
# num de consultas
n = int(input())

p_ent = []   # pila entrada
p_sal = []   # pila salida

for i in range(n):
    cons = input().split()

    if cons[0] == '1':
        # enqueue
        val = int(cons[1])
        p_ent.append(val)

    elif cons[0] == '2':
        # dequeue
        if not p_sal:
            while p_ent:
                p_sal.append(p_ent.pop())
        p_sal.pop()

    elif cons[0] == '3':
        # mostrar frente
        if not p_sal:
            while p_ent:
                p_sal.append(p_ent.pop())
        print(p_sal[-1])
```

## Pruebas funcionales



```python
def procesar_consultas(consultas):
    p_ent = []
    p_sal = []
    resultados = []

    for cons in consultas:
        if cons[0] == 1:
            p_ent.append(cons[1])

        elif cons[0] == 2:
            if not p_sal:
                while p_ent:
                    p_sal.append(p_ent.pop())
            p_sal.pop()

        elif cons[0] == 3:
            if not p_sal:
                while p_ent:
                    p_sal.append(p_ent.pop())
            resultados.append(p_sal[-1])

    return resultados
```

```python
consultas = [(1,42),(2,),(1,14),(3,),(1,28)]
assert procesar_consultas(consultas) == [14]

consultas = [(1,10),(1,20),(3,),(2,),(3,),(2,)]
assert procesar_consultas(consultas) == [10,20]

consultas = [(1,5),(1,6),(1,7),(3,),(2,),(3,),(2,)]
assert procesar_consultas(consultas) == [5,6]

consultas = [(1,1),(1,2),(2,),(1,3),(3,),(2,),(3,),(2,)]
assert procesar_consultas(consultas) == [2,3]

consultas = [(1,100),(1,200),(1,300),(3,),(2,),(3,),(1,400),(2,),(3,),(2,)]
assert procesar_consultas(consultas) == [100,200,300]
```

##PROBLEMA

Knight in a wr grid

##ANALISIS DE COMPLEJIDAD
o(n**3)


##CODIGO

```python
from collections import deque

T = int(input())

for numero_caso in range(1, T + 1):
    filas, columnas, mov_h, mov_v = map(int, input().split())

    cantidad_agua = int(input())

    celdas_agua = []
    for i in range(cantidad_agua):
        x, y = map(int, input().split())
        celdas_agua.append((x, y))

    visitado = [[False for j in range(columnas)] for i in range(filas)]

    cola = deque()
    cola.append((0, 0))
    visitado[0][0] = True

    cantidad_pares = 0
    cantidad_impares = 0

    while cola:
        fila_actual, col_actual = cola.popleft()

        movimientos_posibles = [
            (mov_h, mov_v), (mov_h, -mov_v), (-mov_h, mov_v), (-mov_h, -mov_v),
            (mov_v, mov_h), (mov_v, -mov_h), (-mov_v, mov_h), (-mov_v, -mov_h)
        ]

        vecinos_validos = []

        for mov_x, mov_y in movimientos_posibles:
            nueva_fila = fila_actual + mov_x
            nueva_col = col_actual + mov_y

            # verificar que esté dentro del tablero
            if 0 <= nueva_fila < filas and 0 <= nueva_col < columnas:

                # verificar que no sea agua
                if (nueva_fila, nueva_col) not in celdas_agua:

                    # evitar contar vecinos repetidos
                    if (nueva_fila, nueva_col) not in vecinos_validos:
                        vecinos_validos.append((nueva_fila, nueva_col))

                    # BFS: visitar si no ha sido visitado
                    if not visitado[nueva_fila][nueva_col]:
                        visitado[nueva_fila][nueva_col] = True
                        cola.append((nueva_fila, nueva_col))

        cantidad_vecinos = len(vecinos_validos)

        if cantidad_vecinos % 2 == 0:
            cantidad_pares += 1
        else:
            cantidad_impares += 1

    print("Case", numero_caso, ":", cantidad_pares, cantidad_impares)
```

##PRUEBAS FUNCIONALES

```python
from collections import deque

def contar_celdas(filas, columnas, mov_h, mov_v, celdas_agua):
    visitado = [[False for j in range(columnas)] for i in range(filas)]

    cola = deque()
    cola.append((0, 0))
    visitado[0][0] = True

    pares = 0
    impares = 0

    while cola:
        fila_actual, col_actual = cola.popleft()

        movimientos = [
            (mov_h, mov_v), (mov_h, -mov_v), (-mov_h, mov_v), (-mov_h, -mov_v),
            (mov_v, mov_h), (mov_v, -mov_h), (-mov_v, mov_h), (-mov_v, -mov_h)
        ]

        vecinos = []

        for dx, dy in movimientos:
            nf = fila_actual + dx
            nc = col_actual + dy

            if 0 <= nf < filas and 0 <= nc < columnas:
                if (nf, nc) not in celdas_agua:

                    if (nf, nc) not in vecinos:
                        vecinos.append((nf, nc))

                    if not visitado[nf][nc]:
                        visitado[nf][nc] = True
                        cola.append((nf, nc))

        k = len(vecinos)

        if k % 2 == 0:
            pares += 1
        else:
            impares += 1

    return pares, impares
```

```python
assert contar_celdas(1, 1, 1, 1, []) == (1, 0)

assert contar_celdas(4, 4, 1, 2, []) == (8, 8)

agua = [(1,1), (2,2)]
assert contar_celdas(4, 4, 1, 2, agua) == (6, 6)

assert contar_celdas(3, 3, 1, 1, []) == (4, 0)

assert contar_celdas(3, 3, 0, 1, []) == (4, 0)
```
